# Replay de una run existente usando `evac_sim.viz.animation`

Este notebook reproduce una ejecución ya guardada sin volver a simular.

A diferencia de la versión anterior, aquí se usa el mismo enfoque que en tu `Main.ipynb`:

```python
from evac_sim.viz.animation import animate
```

Además, el `walkable_area` se obtiene desde el entorno configurado en `config_resolved.yaml` mediante:

```python
from evac_sim.envs.environment_factory import select_environment
```

Esto evita depender de que cada SQLite de modo contenga la tabla `geometry`.


````python
from pathlib import Path
import json
import sqlite3
import sys

import pandas as pd
import yaml
import pedpy
````

````python
# ---------------------------------------------------------------------
# User configuration
# ---------------------------------------------------------------------

def find_project_root(start: Path | None = None) -> Path:
    current = (start or Path.cwd()).resolve()

    for parent in [current, *current.parents]:
        if (parent / ".git").exists():
            return parent
        if (parent / "src").exists() and (parent / "runs").exists():
            return parent

    return current


PROJECT_ROOT = find_project_root()

# Change this to the run directory you want to replay.
#RUN_DIR = PROJECT_ROOT / "runs" / "profile_h1_short_vs_wide_cprofile"
# RUN_DIR = PROJECT_ROOT / "runs"/"congestion_heuristics_efficient_high_v1"/"h3"/"congestion_short_vs_wide"
RUN_DIR = PROJECT_ROOT / "runs"/"diagnostics_spatial_queue_two_exits"/"congestion_two_exits_h1"
#RUN_DIR = PROJECT_ROOT / "runs"/"diagnostics_local_movement_capacity_two_exits"/"congestion_two_exits_h3"

# Mode used to choose the trajectory database.
# Examples: "none", "h1", "h2", "h3".
MODE = "h3"

# Force a DB if needed. Leave as None for automatic discovery.
# Example:
# DB_PATH = RUN_DIR / "artifacts" / "db" / "h2.sqlite"
DB_PATH = None

# Animation settings.
EVERY_NTH_FRAME = 50
FIG_WIDTH = 900
FIG_HEIGHT = 900
AGENT_RADIUS = 0.2

print(f"PROJECT_ROOT: {PROJECT_ROOT}")
print(f"RUN_DIR:      {RUN_DIR}")
````

````python
# ---------------------------------------------------------------------
# Project imports
# ---------------------------------------------------------------------

SRC_DIR = PROJECT_ROOT / "src"

if str(SRC_DIR) not in sys.path:
    sys.path.insert(0, str(SRC_DIR))

from evac_sim.viz.animation import animate
from evac_sim.envs.environment_factory import ENVIRONMENTS, select_environment

print("Imported:")
print("  animate from evac_sim.viz.animation")
print("  select_environment from evac_sim.envs.environment_factory")
````

````python
# ---------------------------------------------------------------------
# SQLite discovery
# ---------------------------------------------------------------------

def is_sqlite_file(path: Path) -> bool:
    if not path.is_file():
        return False

    if path.suffix.lower() not in {".sqlite", ".sqlite3", ".db", ".squlite"}:
        return False

    try:
        with path.open("rb") as f:
            return f.read(16).startswith(b"SQLite format 3")
    except OSError:
        return False


def sqlite_tables(db_path: Path) -> pd.DataFrame:
    with sqlite3.connect(db_path) as conn:
        return pd.read_sql_query(
            '''
            SELECT name, type
            FROM sqlite_master
            WHERE type IN ('table', 'view')
            ORDER BY type, name
            ''',
            conn,
        )


def sqlite_has_table(db_path: Path, table_name: str) -> bool:
    with sqlite3.connect(db_path) as conn:
        row = conn.execute(
            '''
            SELECT 1
            FROM sqlite_master
            WHERE type IN ('table', 'view')
              AND name = ?
            LIMIT 1
            ''',
            (table_name,),
        ).fetchone()

    return row is not None


def find_sqlite_files(run_dir: Path) -> list[Path]:
    if not run_dir.exists():
        raise FileNotFoundError(
            f"RUN_DIR does not exist: {run_dir}\n"
            f"Current working directory: {Path.cwd()}"
        )

    search_roots = [
        run_dir / "artifacts" / "db",
        run_dir / "artifacts",
        run_dir,
    ]

    files: list[Path] = []

    for search_root in search_roots:
        if search_root.exists():
            files.extend(
                path for path in search_root.rglob("*")
                if is_sqlite_file(path)
            )

    return sorted(set(files))


def choose_trajectory_db(sqlite_files: list[Path], mode: str) -> Path:
    if not sqlite_files:
        raise FileNotFoundError("No SQLite files were found.")

    mode_lower = mode.lower()

    # Prefer a DB whose name contains the selected mode and has trajectory_data.
    mode_candidates = [
        path for path in sqlite_files
        if mode_lower in path.stem.lower()
        and sqlite_has_table(path, "trajectory_data")
    ]

    if mode_candidates:
        return mode_candidates[0]

    # Then any DB with trajectory_data except simulation.db if possible.
    trajectory_candidates = [
        path for path in sqlite_files
        if sqlite_has_table(path, "trajectory_data")
    ]

    non_simulation = [
        path for path in trajectory_candidates
        if path.name.lower() != "simulation.db"
    ]

    if non_simulation:
        return non_simulation[0]

    if trajectory_candidates:
        return trajectory_candidates[0]

    raise FileNotFoundError(
        "SQLite files were found, but none contains the table 'trajectory_data'."
    )


if DB_PATH is None:
    sqlite_files = find_sqlite_files(RUN_DIR)

    print("SQLite files found:")
    for index, path in enumerate(sqlite_files):
        try:
            relative = path.relative_to(PROJECT_ROOT)
        except ValueError:
            relative = path

        marker = "trajectory_data" if sqlite_has_table(path, "trajectory_data") else ""
        print(f"[{index}] {relative} {marker}")

    DB_PATH = choose_trajectory_db(sqlite_files, MODE)
else:
    DB_PATH = Path(DB_PATH)
    if not DB_PATH.is_absolute():
        DB_PATH = PROJECT_ROOT / DB_PATH

if not DB_PATH.exists():
    raise FileNotFoundError(f"DB_PATH does not exist: {DB_PATH}")

print(f"\nUsing trajectory DB_PATH: {DB_PATH}")
display(sqlite_tables(DB_PATH))
````

````python
# ---------------------------------------------------------------------
# Read config_resolved.yaml and get environment name
# ---------------------------------------------------------------------

def find_config_resolved(run_dir: Path) -> Path:
    candidates = [
        run_dir / "config_resolved.yaml",
        run_dir / "artifacts" / "config_resolved.yaml",
        run_dir / "artifacts" / "config" / "config_resolved.yaml",
    ]

    for candidate in candidates:
        if candidate.exists():
            return candidate

    matches = sorted(run_dir.rglob("config_resolved.yaml"))

    if not matches:
        raise FileNotFoundError(
            f"Could not find config_resolved.yaml under {run_dir}"
        )

    return matches[0]


def iter_nested_items(value):
    if isinstance(value, dict):
        for key, child in value.items():
            yield key, child
            yield from iter_nested_items(child)
    elif isinstance(value, list):
        for child in value:
            yield from iter_nested_items(child)


def extract_environment_name(config: dict) -> str:
    # Common locations.
    candidates = [
        config.get("environment"),
        config.get("env"),
        config.get("environment_name"),
    ]

    for section_name in ("case", "simulation", "experiment", "run", "scenario"):
        section = config.get(section_name)

        if isinstance(section, dict):
            candidates.extend([
                section.get("environment"),
                section.get("env"),
                section.get("environment_name"),
            ])

    for candidate in candidates:
        if candidate:
            return str(candidate)

    # Fallback recursive search.
    for key, value in iter_nested_items(config):
        if str(key).lower() in {"environment", "env", "environment_name"} and value:
            return str(value)

    raise KeyError(
        "Could not find environment/env/environment_name in config_resolved.yaml."
    )


def normalize_environment_name(environment_name: str) -> str:
    aliases = {
        "congestion_short_vs_wide": "short_vs_wide",
        "congestion_parallel_corridors": "parallel_corridors",
        "congestion_two_exits": "two_exits",
    }

    candidate = aliases.get(environment_name, environment_name)

    if candidate in ENVIRONMENTS:
        return candidate

    if candidate.startswith("congestion_"):
        stripped = candidate.removeprefix("congestion_")
        if stripped in ENVIRONMENTS:
            return stripped

    return candidate


CONFIG_RESOLVED_PATH = find_config_resolved(RUN_DIR)

with CONFIG_RESOLVED_PATH.open("r", encoding="utf-8") as f:
    RESOLVED_CONFIG = yaml.safe_load(f) or {}

RAW_ENVIRONMENT_NAME = extract_environment_name(RESOLVED_CONFIG)
ENVIRONMENT_NAME = normalize_environment_name(RAW_ENVIRONMENT_NAME)

print(f"CONFIG_RESOLVED_PATH: {CONFIG_RESOLVED_PATH}")
print(f"RAW_ENVIRONMENT_NAME: {RAW_ENVIRONMENT_NAME}")
print(f"ENVIRONMENT_NAME:     {ENVIRONMENT_NAME}")

if ENVIRONMENT_NAME not in ENVIRONMENTS:
    raise ValueError(
        f"Unknown environment '{ENVIRONMENT_NAME}'. "
        f"Available: {sorted(ENVIRONMENTS)}"
    )
````

````python
# ---------------------------------------------------------------------
# Load walkable area from environment_factory
# ---------------------------------------------------------------------

def to_pedpy_walkable_area(value):
    if value is None:
        return None

    if isinstance(value, pedpy.WalkableArea):
        return value

    # Shapely geometry.
    if hasattr(value, "wkt"):
        return pedpy.WalkableArea(value.wkt)

    # WKT string.
    if isinstance(value, str):
        text = value.strip()
        if text.upper().startswith(("POLYGON", "MULTIPOLYGON")):
            return pedpy.WalkableArea(text)

    # Object attributes commonly used for environments.
    for attr_name in (
        "walkable_area",
        "area",
        "walkable",
        "geometry",
        "polygon",
    ):
        if hasattr(value, attr_name):
            converted = to_pedpy_walkable_area(getattr(value, attr_name))
            if converted is not None:
                return converted

    # Dictionaries.
    if isinstance(value, dict):
        for key in (
            "walkable_area",
            "area",
            "walkable",
            "geometry",
            "polygon",
        ):
            if key in value:
                converted = to_pedpy_walkable_area(value[key])
                if converted is not None:
                    return converted

        for child in value.values():
            converted = to_pedpy_walkable_area(child)
            if converted is not None:
                return converted

    # Tuples/lists returned by environment builders.
    if isinstance(value, (tuple, list)):
        for child in value:
            converted = to_pedpy_walkable_area(child)
            if converted is not None:
                return converted

    return None


ENVIRONMENT_OBJECT = select_environment(ENVIRONMENT_NAME)
WALKABLE_AREA = to_pedpy_walkable_area(ENVIRONMENT_OBJECT)

if WALKABLE_AREA is None:
    raise ValueError(
        f"Could not extract walkable_area from environment '{ENVIRONMENT_NAME}'. "
        "Inspect ENVIRONMENT_OBJECT to see its structure."
    )

print(f"Walkable area loaded from environment '{ENVIRONMENT_NAME}'.")
print(type(WALKABLE_AREA))
````

````python
# ---------------------------------------------------------------------
# Read trajectory data from selected SQLite file
# ---------------------------------------------------------------------

def read_trajectory_from_sqlite(db_path: Path) -> pedpy.TrajectoryData:
    with sqlite3.connect(db_path) as con:
        data = pd.read_sql_query(
            '''
            SELECT
                frame,
                id,
                pos_x AS x,
                pos_y AS y,
                ori_x AS ox,
                ori_y AS oy
            FROM trajectory_data
            ORDER BY frame, id
            ''',
            con,
        )

        fps_row = con.cursor().execute(
            "SELECT value FROM metadata WHERE key = 'fps'"
        ).fetchone()

    fps = float(fps_row[0]) if fps_row is not None else 10.0

    return pedpy.TrajectoryData(
        data=data,
        frame_rate=fps,
    )


TRAJECTORY_DATA = read_trajectory_from_sqlite(DB_PATH)

display(TRAJECTORY_DATA.data.head())

print(f"Rows:   {len(TRAJECTORY_DATA.data):,}")
print(f"Frames: {TRAJECTORY_DATA.data['frame'].min()} -> {TRAJECTORY_DATA.data['frame'].max()}")
print(f"Agents: {TRAJECTORY_DATA.data['id'].nunique()}")
print(f"FPS:    {TRAJECTORY_DATA.frame_rate}")
````

````python
# ---------------------------------------------------------------------
# Animate stored run
# ---------------------------------------------------------------------

fig = animate(
    TRAJECTORY_DATA,
    WALKABLE_AREA,
    every_nth_frame=EVERY_NTH_FRAME,
    width=FIG_WIDTH,
    height=FIG_HEIGHT,
    radius=AGENT_RADIUS,
    title_note=f"{ENVIRONMENT_NAME} | {DB_PATH.name}",
)

fig.show()
````
